```python
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

from src.data import load_features, time_split, FEATURE_COLS
from src.lr import LinearRegressionScratch
```

```python
feat = load_features()
train_sets, test_sets = time_split(feat)
```

```python
tr, te = train_sets["TCS.NS"], test_sets["TCS.NS"]
scaler = StandardScaler()
Xtr = scaler.fit_transform(tr[FEATURE_COLS])
Xte = scaler.transform(te[FEATURE_COLS])
ytr = tr["target"].values
yte = te["target"].values
```

```python
mine = LinearRegressionScratch()
mine.fit(Xtr, ytr)
sk = LinearRegression().fit(Xtr, ytr)
```

```python
print("max weight difference:", np.abs(mine.w - sk.coef_).max())
print("max prediction difference:", np.abs(mine.predict(Xte) - sk.predict(Xte)).max())
```
